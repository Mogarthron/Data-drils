```python
import pandas as pd
```

```python
df = pd.read_csv("hotel_bookings.csv", parse_dates=["checkin_date", "checkout_date"])
df
```

```python
df = df[df.is_canceled == 0]
df["nights"] = [pd.date_range(ci, co, inclusive='left') for ci, co in zip(df.checkin_date, df.checkout_date)]
```

```python
df_month = (
    df.explode("nights").nights.dt.to_period("M")
    .value_counts()
    .sort_index()
    .to_frame("boooked_nghts")
    .assign(
        avaible_nights = lambda x: x.index.days_in_month * 200,
        oqupacy = lambda x: x["boooked_nghts"] / x["avaible_nights"]
    )
)
```
