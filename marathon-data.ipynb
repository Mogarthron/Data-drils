```python
import pandas as pd
```

```python
df = pd.read_csv("marathon-data.csv")
df
```

```python
def time_bands(x):
    if x < 180:
        return "sub 3"
    elif x >= 180 and x < 210: 
        return "3:00 - 3:30"
    elif x >= 210 and x < 240:
        return "3:30 - 4:00"
    elif x >= 240 and x < 270:
        return "4:00 - 4:30"
    elif x >= 270 and x < 300:
        return "4:30 - 5:00"
    elif x >= 300 and x < 330:
        return "5:00 - 5:30"
    elif x >= 330 and x < 360:
        return "5:30 - 6:00"
    return "6 +"

def age_bands(x):
    if x < 20:
        return "sub 20"
    elif x >= 20 and x < 30: 
        return "20 - 30"
    elif x >= 30 and x < 40:
        return "30 - 40"
    elif x >= 40 and x < 50:
        return "40 - 50"
    elif x >= 50 and x < 60:
        return "50 - 60"
    elif x >= 60 and x < 70:
        return "60 - 70"
    return "70 +"

df = (
    df.assign(
        minutes = lambda x: pd.to_timedelta(x["final"]).dt.total_seconds() / 60,
        time_bands = lambda x: x["minutes"].apply(time_bands),
        age_bands = lambda x: x["age"].apply(age_bands)
    )
)

#df.value_counts("age", normalize=True)
```

```python
df.groupby(["gender","age_bands","time_bands"]).agg(
   time_prc = ("time_bands", lambda x: x.count() / df.shape[0]) 
)
```

```python
df.groupby("age_bands").agg(
   age_prc = ("age_bands", lambda x: x.count() / df.shape[0]) 
)
```

```python
df.groupby("gender").agg(
   gender_prc = ("gender", lambda x: x.count() / df.shape[0]) 
)
```
